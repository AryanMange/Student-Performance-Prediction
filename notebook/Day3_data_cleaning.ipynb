```python
import os
import pandas as pd
import numpy as np

```

```python
data_path = os.path.join("..", "data", "student_data.csv")
if not os.path.exists(data_path):
    data_path = os.path.join("data", "student_data.csv")
if not os.path.exists(data_path):
    data_path = "student_data.csv"

df = pd.read_csv(data_path)
print(f"Initial shape: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
num_duplicates = df.duplicated().sum()
print(f"Total duplicate rows found: {num_duplicates}")
```

```python
print("--- Missing Values Before Cleaning ---")
nulls_before = df.isnull().sum()
print(nulls_before[nulls_before > 0])
```

```python
cols_to_impute = ['Teacher_Quality', 'Parental_Education_Level', 'Distance_from_Home']

for col in cols_to_impute:
    col_mode = df[col].mode()[0]
    print(f"Imputing '{col}' missing values with mode: '{col_mode}'")
    df[col] = df[col].fillna(col_mode)
```

```python
print("\n--- Missing Values After Cleaning ---")
print(f"Total missing values remaining: {df.isnull().sum().sum()}")
```

```python
print(f"Hours Studied range: {df['Hours_Studied'].min()} to {df['Hours_Studied'].max()}")
print(f"Attendance range:    {df['Attendance'].min()}% to {df['Attendance'].max()}%")
print(f"Previous Score range:{df['Previous_Scores'].min()} to {df['Previous_Scores'].max()}")
print(f"Exam Score range:    {df['Exam_Score'].min()} to {df['Exam_Score'].max()}")
```

```python
anomaly_mask = df['Exam_Score'] > 100
print(f"\nNumber of records with Exam_Score > 100: {anomaly_mask.sum()}")
if anomaly_mask.sum() > 0:
    display(df[anomaly_mask])
```

```python
df.loc[anomaly_mask, 'Exam_Score'] = 100
print("Cap applied: Exam_Score capped at 100.")
```

```python
clean_data_path = os.path.join("..", "data", "student_data_cleaned.csv")
if not os.path.exists(os.path.dirname(clean_data_path)):
    clean_data_path = os.path.join("data", "student_data_cleaned.csv")


target_dir = os.path.dirname(clean_data_path)
if target_dir and not os.path.exists(target_dir):
    os.makedirs(target_dir, exist_ok=True)

df.to_csv(clean_data_path, index=False)
print(f"Cleaned dataset saved successfully to: {clean_data_path}")
print(f"Final clean dataset shape: {df.shape}")
```
