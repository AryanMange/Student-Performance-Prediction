```python
import os
import pandas as pd
import numpy as np
```

```python
data_path = os.path.join("..", "data", "student_data.csv")
if not os.path.exists(data_path):
    data_path = os.path.join("data", "student_data.csv")
if not os.path.exists(data_path):
    data_path = "student_data.csv"

df = pd.read_csv(data_path)
```

```python
print("--- Unique Categories per Column ---")
for col in df.select_dtypes(include=['object']).columns:
    unique_vals = list(df[col].dropna().unique())
    print(f"{col:<28}: {len(unique_vals)} unique -> {unique_vals}")
```

```python
null_counts = df.isnull().sum()
missing_df = pd.DataFrame({
    'Missing_Count': null_counts[null_counts > 0],
    'Percentage (%)': (null_counts[null_counts > 0] / len(df) * 100).round(2)
})

print("--- Columns with Missing Values ---")
display(missing_df)
```
